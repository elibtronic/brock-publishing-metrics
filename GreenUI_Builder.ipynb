- OpenAlex search
- time period: `2025-07-01 - 2026-06-30`
- affilation via: `authorships.institutions.lineage:i186803428`
- search string: `(green OR environment OR sustainability OR renewable energy OR climate change)`


via [web](https://openalex.org/works?page=1&filter=authorships.institutions.lineage:i186803428,from_publication_date:2025-07-01,to_publication_date:2026-06-30,type:article&sort=relevance_score:desc&search.title_and_abstract=(green+OR+environment+OR+sustainability+OR+renewable+energy+OR+climate+change))

```python
import requests
import numpy as np
import json
import pandas as pd
from google.colab import files
from pprint import pprint

from_publication_date = "2025-07-01"  # @param {type:"date"}
to_publication_date = "2026-06-30"  # @param {type:"date"}
api_key = "KEY_HERE"  # @param {type:"string"}
file_name = "green_ui_2025" # @param {type:"string"}


```

```python
#filter = f"authorships.institutions.ror:{ror_id},from_publication_date:{from_publication_date},to_publication_date:{to_publication_date},type:article,primary_location.source.type:journal"
filter = f"authorships.institutions.lineage:i186803428,from_publication_date:{from_publication_date},to_publication_date:{to_publication_date},type:article&sort=relevance_score:desc&search.title_and_abstract=(green+OR+environment+OR+sustainability+OR+renewable+energy+OR+climate+change)"

api_url = f"https://api.openalex.org/works?filter={filter}&per-page=200&api_key={api_key}"

cursor = "*"  # Cursor for the first request
works = []  # List to accumulate all results

while cursor:
    # Construct the request URL with the current cursor
    request_url = api_url + f"&cursor={cursor}"

    # Get results from the API
    page = requests.get(request_url).json()
    results = page.get('results')
    meta = page.get('meta')

    # Accumulate results
    works += results

    # Get the next cursor from the current results
    cursor = meta.get('next_cursor')

    # Print progress
    total = int(meta.get("count"))
    #print(f"\r{len(works)}/{total}: {request_url}", end="")

# Do something with all results
print(f"\nTotal works: {len(works)}")

api_url = f"https://api.openalex.org/works?filter={filter}&per-page=200&api_key={api_key}"

cursor = "*"  # Cursor for the first request
works = []  # List to accumulate all results

while cursor:
    # Construct the request URL with the current cursor
    request_url = api_url + f"&cursor={cursor}"

    # Get results from the API
    page = requests.get(request_url).json()
    results = page.get('results')
    meta = page.get('meta')

    # Accumulate results
    works += results

    # Get the next cursor from the current results
    cursor = meta.get('next_cursor')

    # Print progress
    total = int(meta.get("count"))
    #print(f"\r{len(works)}/{total}: {request_url}", end="")

# Do something with all results
print(f"\nTotal works: {len(works)}")

#Make DF of results
g_df = pd.json_normalize(results)
```
