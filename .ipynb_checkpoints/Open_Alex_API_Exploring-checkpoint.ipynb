```python
def journal_details(issn):   
    detail_string = ""
    try:
        j_item = requests.get("https://api.openalex.org/sources/issn:"+issn).json()
        if not j_item['is_oa']:
            detail_string += "<br>Title is _closed_ and might be eligible for discount. Please see [journal homepage]("+j_item['homepage_url']+")"
        else: 
            detail_string += "<br>Could not determine if title is eligible for discount, or title might already be open access. Please see [journal homepage]("+j_item['homepage_url']+")"
        
        if j_item["apc_usd"]: 
            detail_string += "<br>Usual Author Processing Charge for this title is: **"+str(j_item["apc_usd"])+"** USD"

        detail_string += "<br>More analytics for this title from [OpenAlex]("+j_item['ids']['openalex']+")"
    except:
        detail_string = "**Could not retrieve journal information**"
    return detail_string
    

issns_to_check = [
    "0010-9452",
    "0887-8730",
    "2047-4849",
]


for i in issns_to_check:
    display(Markdown("**"+i+"**"), Markdown(journal_details(i)))
```
